```python

# Question: Implement similarity search using FAISS

# Step 1: Install required libraries
!pip install -q faiss-cpu numpy

# Step 2: Import libraries
import faiss
import numpy as np

# Step 3: Create sample documents
documents = [
    "Artificial Intelligence enables machines to perform intelligent tasks.",
    "Machine Learning allows computers to learn from data.",
    "Deep Learning uses neural networks for complex tasks.",
    "Python is a popular programming language."
]

# Step 4: Create sample embeddings (2-dimensional vectors)
vectors = np.array([
    [1.0, 0.0],
    [0.9, 0.1],
    [0.0, 1.0],
    [0.1, 0.9]
], dtype="float32")

# Step 5: Create a FAISS index
dimension = vectors.shape[1]
index = faiss.IndexFlatL2(dimension)

# Step 6: Add document vectors to the index
index.add(vectors)

# Step 7: Create a query vector
query = np.array([[1.0, 0.1]], dtype="float32")

# Step 8: Search for the 2 nearest documents
k = 2
distances, indices = index.search(query, k)

# Step 9: Display retrieved documents
print("User Query: Find documents related to AI\n")
print("Top 2 Similar Documents:\n")

for rank, idx in enumerate(indices[0]):
    print(f"Rank: {rank + 1}")
    print(f"Document: {documents[idx]}")
    print(f"L2 Distance: {distances[0][rank]:.4f}")
    print()
```
