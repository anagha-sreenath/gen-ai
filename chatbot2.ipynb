```python

# MINI GENERATIVE AI CHATBOT

knowledge_base = {
    "generative ai": "Generative AI is a type of artificial intelligence that creates new content such as text, images, audio and code.",
    "large language model": "An LLM is a Large Language Model trained on large amounts of text to understand and generate human-like language.",
    "rag": "RAG stands for Retrieval-Augmented Generation. It retrieves relevant information from external documents and uses it to generate accurate answers.",
    "prompt engineering": "Prompt engineering is the process of designing effective instructions to get better responses from AI models.",
    "machine learning": "Machine Learning is a branch of AI that enables computers to learn patterns from data and make predictions.",
    "deep learning": "Deep Learning is a subset of machine learning that uses neural networks with multiple layers.",
    "hallucination": "AI hallucination occurs when an AI model generates incorrect or fabricated information.",
    "transformer": "A Transformer is a neural network architecture that uses attention mechanisms to process sequential data."
}


def chatbot():
    print("=" * 45)
    print("       MINI GENAI CHATBOT")
    print("=" * 45)
    print("Ask me anything about GenAI!")
    print("Type 'exit' to stop.\n")

    while True:
        user_input = input("You: ").lower().strip()

        if user_input == "exit":
            print("Bot: Goodbye! Have a great day.")
            break

        if not user_input:
            print("Bot: Please enter a question.")
            continue

        found = False

        for keyword, answer in knowledge_base.items():
            if keyword in user_input:
                print("Bot:", answer)
                found = True
                break

        if not found:
            print("Bot: I don't know the answer to that question.")


chatbot()
```
